```python
import pandas as pd

scores = pd.Series([45, 55, 65, 75, 85, 95])

result = pd.cut(
    scores,
    bins=[0, 60, 80, 100],       # 구간의 경계값
    labels=["하", "중", "상"]    # 각 구간에 붙일 이름
    # 0 ───── 60 ───── 80 ───── 100
    #     하       중       상
)

display(result)
```

```python
result = pd.cut(
    scores,
    bins=[0, 60, 80, 100],
    right=True # 오른쪽 구간 포함 -기본값
)

display(result)
```

```python
scores = pd.Series([45, 55, 65, 75, 85, 95])
result = pd.cut(
    scores,
    bins=[0, 60, 80, 100],
    right=False
)

display(result)
```
