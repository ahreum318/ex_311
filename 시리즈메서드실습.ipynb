```python
# 구간별 값을 매핑하는 메서드 x
# apply(함수) 메서드: apply 메서드 매개값으로 들어갈 함수 선언(람다 사용 가능)

def 매출_등급(val):
    if pd.isna(val): #val은 시리즈가 아님. val.isna() X, pd.isna(값) 함수 사용
        return "결측치"
    elif val >= 1000:
        return "상"
    elif val >= 800:
        return "중"
    else:
        return "하"

등급 = 매출.apply(매출_등급)
등급
```
