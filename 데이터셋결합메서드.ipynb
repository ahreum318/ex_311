```python
import pandas as pd

df1 = pd.DataFrame({
    "이름": ["홍길동", "김철수"],
    "점수": [80, 90]
})

df2 = pd.DataFrame({
    "이름": ["이영희", "박민수"],
    "점수": [85, 95]
})

display(df1)
display(df2)
```

```python
df3 = pd.concat([df1, df2], axis=0, ignore_index=True)
df3
```

```python
df4 = pd.concat([df1, df2], axis=1, ignore_index=False)
df4
```

```python
# axis=0 위아래 결합(기본값) -> 행추가
df = pd.concat([df1, df2], ignore_index=False, axis=0) # ignore_index 기본값 False
display(df)
```

```python
df = pd.concat([df1, df2], ignore_index=True, axis=0)
display(df)
```

```python
# merge 함수: SQL JOIN 유사
import pandas as pd

df_customer = pd.DataFrame({
    "고객ID": [101, 102, 103, 104],
    "이름": ["홍길동", "김철수", "이영희", "박민수"]
})

df_score = pd.DataFrame({
    "고객ID": [101, 102, 103, 105],
    "점수": [90, 80, 95, 70]
})


display(df_customer)
display(df_score)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="inner" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="outer" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="left" # inner(기본값), left, right, outer
)

display(result)
```

```python
result = pd.merge(
    df_customer,
    df_score,
    on="고객ID",
    how="right" # inner(기본값), left, right, outer
)

display(result)
```
