```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df)
```

```python
df["이름"]
```

```python
# 새로운 열 = 시리즈 타입
# 리스트 -> 시리즈 변경 후 대입
df["성적"] = ['B','A','C','A+','B+']
df
```

```python
df["생년"] = 2000 # = 값 -> 같은 값의 시리즈로 변경 후 입력
df
```

```python
df["가입년도"] = [2010, 2009, 2011] # 수량이 다르면 에러
df
```

```python
df
```

```python
r = df['점수'] + df['출석'] # 시리즈끼리 더하면 시리즈로 나옴
print(r)
print(type(r))
```

```python
df['총점'] = df['점수'] + df['출석']

display(df)
```

```python
df["생년"] = 1999 # 1999 스칼라 값이 Series[1999,1999,1999,1999,1999]동일한 사이즈의 시리즈로 변경 후 대입
df
```

```python
df.iloc[2, 4] = "C+"
df
```

```python
df.loc[2, "성적"] ="C"
df
```

```python
df.loc[[2,4], '출석'] = 88
```

```python
df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df.count())
print(df.sum(numeric_only=True))
print(df.mean(numeric_only=True))
print(df.median(numeric_only=True))
print(df.min(numeric_only=True))
print(df.max(numeric_only=True))
print(df.std(numeric_only=True))
print(df.std(numeric_only=True))
print(df['점수'].max() - df['점수'].min())
```

```python
df.isnull()
# 데이터 행이나 열이 많으면 사용 불가능
# 데이터가 작은 경우에만 사용가능
```

```python
df.isnull().sum() #boolean 타입의 sum()은 False: 0, True: 1 계산
```

```python
# 열만 추출 데이터셋
# df["gender"] 추출된 결과 아래와 같다

gender = pd.Series([
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "M"
])

print(gender.describe())
print(gender.unique())
print(gender.nunique())
print(gender.value_counts())

g_bar = gender.value_counts()
#g_bar.plot(kind="bar")
gender.value_counts().plot(kind="bar")

# 인과관계는 선차트
# 상관관계는 산점도
```
