```python
import pandas as pd
```

```python
# {레이블: [], 레이블: [] ...}
df1 = pd.DataFrame({
                    "이름":["민수","지영"],
                    "점수":[90,85]
                    })

# print() #터미널출력
# display() #노트북 셀에 최적화 출력

print(df1)
display(df1)
```

```python
#index값 행 레이블(레이블 인덱스), columns 값 열 레이블(레이블 인덱스)
df2 = pd.DataFrame([
    ["민수", 90], #1행
    ["지영", 85], #2행
], index = ["첫번째","두번째"], columns = ["이름", "점수"]
)

df2
```

```python
df3 = pd.DataFrame([
    {"이름":"민수", "점수":90},
    {"이름":"지영", "점수":100}
])

df3
```

```python
# numpy의 2차원 배열 - 모든 데이터 타입이 동일
import numpy as np

df3 = pd.DataFrame(np.array([
    [1, 2], 
    [3, 4]
]), columns = ["x", "y"])

df3
```
