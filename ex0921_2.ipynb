```python
import matplotlib.pyplot as plt
import seaborn as sns

# mpg 데이터 불러오기
mpg = sns.load_dataset("mpg")

# 데이터 확인
print(mpg.head())

# 숫자형 변수들의 상관계수 계산
corr = mpg.corr(numeric_only=True)

# 상관계수 행렬 확인
print(corr)

# Figure 생성
fig = plt.figure(figsize=(10, 8))

# Figure 안에 1개의 Axes 생성
axes1 = fig.add_subplot(1, 1, 1)

# 상관계수 행렬을 Heatmap으로 표현
sns.heatmap(
    corr,
    annot=True,       # 각 칸에 상관계수 숫자 표시
    fmt=".2f",        # 소수점 둘째 자리까지 표시
    ax=axes1
)

# 제목 설정
axes1.set_title("MPG Data Correlation Heatmap")

# 그래프 여백 정리
fig.tight_layout()

# 그래프 출력
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# 한글 폰트 설정 (Windows 기준)
plt.rc('font', family='Malgun Gothic')
plt.rc('axes', unicode_minus=False)

# 데이터셋 로드
df = sns.load_dataset("penguins")
```

```python
# 1. 기술통계량 확인

print("=== 1-1. 기술통계량 (수치형 데이터) ===")
display(df.describe())

print("=== 1-2. 기술통계량 (범주형 데이터) ===")
display(df.describe(include=['object', 'category']))
```

```python
# 2. 결측치 확인 및 삭제 (인덱스 3, 339 삭제)
print("=== 2-1. 결측치 개수 ===")
display(df.isnull().sum().to_frame(name='결측치 수'))

# 인덱스 3과 339 행 삭제
df_clean = df.drop(index=[3, 339]).reset_index(drop=True)

print("=== 2-2. 삭제 후 데이터 크기(Shape) ===")
display(pd.DataFrame([df_clean.shape], columns=['행(Rows)', '열(Columns)']))
```

```python
# 3. 분포 확인 (평균, 중앙값) 및 species 분포 확인
numeric_cols = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']

# 전체 수치형 변수별 평균 및 중앙값 비교
mean_median_df = pd.DataFrame({
    '평균(Mean)': df_clean[numeric_cols].mean(),
    '중앙값(Median)': df_clean[numeric_cols].median()
})

print("=== 3-1. 수치형 데이터 전체 평균 vs 중앙값 ===")
display(mean_median_df)

# species(종)별 평균 및 중앙값
print("=== 3-2. 종(species)별 수치 데이터 평균 및 중앙값 ===")
species_stats = df_clean.groupby('species')[numeric_cols].agg(['mean', 'median'])
display(species_stats)

# species(종) 개수 및 비율 분포 확인
species_counts = df_clean['species'].value_counts()
species_pct = df_clean['species'].value_counts(normalize=True) * 100
species_df = pd.DataFrame({'개수': species_counts, '비율(%)': species_pct})

print("=== 3-3. 종(species)별 개수 및 비율 ===")
display(species_df)

# 종(species) 빈도 시각화
plt.figure(figsize=(6, 4))
sns.countplot(data=df_clean, x='species', palette='Set2')
plt.title('펭귄 종(Species)별 개수 분포')
plt.xlabel('종(Species)')
plt.ylabel('개수')
plt.show()
```

```python
# 4. 상관관계 확인 (수치 및 시각화)
# 수치형 데이터 간 상관계수 계산
corr_matrix = df_clean[numeric_cols].corr()

print("=== 4-1. 수치형 데이터 간 상관계수 (corr) ===")
display(corr_matrix)

# 상관관계 히트맵 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title('펭귄 신체 치수 간 상관관계 히트맵')
plt.show()

# 주요 변수 간 산점도 그래프 (종별 구분)
sns.pairplot(df_clean, hue='species', vars=numeric_cols, palette='Set2')
plt.suptitle('펭귄 변수 간 산점도 및 분포 (종별)', y=1.02)
plt.show()
```
