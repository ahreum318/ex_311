```python
text = " ".join(df["opinion"])
print(text)
```

```python
from collections import Counter

# 수치값으로 확인
words = text.lower().split()
print(words)
word_count = Counter(words)
word_count
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=set()       # 기본 불용어 사용 안 함
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```

```python
stopwords = {
    "the", "and", "is", "to", "was",
    "a", "of", "in", "with", "me",
    "my", "more", "than", "I"
}

wordcloud = WordCloud(
    width=800,
    height=500,
    background_color="white",
    stopwords=stopwords # 불용어 설정
).generate(text)

plt.figure(figsize=(10, 6))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
